### Validate the stored Frobenius traces

For each coefficient tuple modulo $p$, use the same integers as a Weierstrass equation over $\mathbb{Q}$. If that equation is singular over $\mathbb{Q}$, change $a_6$ by $p$ or $-p$; this preserves the equation modulo $p$. These three choices suffice because the discriminant is a nonzero quadratic polynomial in $a_6$. The local trace `E.ap(p)` therefore independently checks the table entry.

```python
def elliptic_curve_lift(coeffs, p):
    """Return a nonsingular curve over Q with these coefficients mod p."""
    coeffs = tuple(ZZ(c) for c in coeffs)
    for delta in (0, p, -p):
        lifted_coeffs = coeffs[:-1] + (coeffs[-1] + delta,)
        try:
            E = EllipticCurve(QQ, lifted_coeffs)
        except ArithmeticError:
            continue
        assert all((a - c) % p == 0 for a, c in zip(E.a_invariants(), coeffs))
        return E
    raise AssertionError(f"Could not find a nonsingular lift of {coeffs} mod {p}")


def validate_trace_table(data, p, trace_key):
    mismatches = []
    adjusted = 0
    lifts = {}

    for coeffs, row in data.items():
        E = elliptic_curve_lift(coeffs, p)
        lift_coeffs = tuple(E.a_invariants())
        lifts[coeffs] = lift_coeffs
        adjusted += lift_coeffs != tuple(coeffs)
        computed_trace = ZZ(E.ap(p))
        if computed_trace != row[trace_key]:
            mismatches.append((coeffs, lift_coeffs, row[trace_key], computed_trace))

    assert not mismatches, (
        f"{trace_key} mismatches (mod-p coefficients, lift, stored, computed): "
        f"{mismatches}"
    )
    print(
        f"Verified all {len(data)} {trace_key} values using Q-lifts; "
        f"{adjusted} singular rational models required an a6 adjustment."
    )
    return lifts


assert set(mod2data) == set(product((0, 1), repeat=5))
mod2lifts = validate_trace_table(mod2data, 2, "a2")
```

### Closed-form predictors

These are the formulas checked exhaustively in the following cells and against the CSV dataset at the end of the notebook.

```python
def compute_w1(a2):
    return int(a2) % 2


def compute_w2(a2, a3):
    a2_mod2 = int(a2) % 2
    return (int(a3) + 1 - a2_mod2) % 3 - 1


def compute_w3(N, a2, w2):
    N_mod2 = int(N) % 2
    a2 = int(a2)
    if a2 % 2 == 0:
        return N_mod2
    return (N_mod2 + 1 + int(w2) + (a2 + 1) // 2) % 2
```

## Extra sanity check against the downloaded dataset

The finite checks above prove the formulas on the relevant residue classes. As a second, independent source of ground truth, check the formulas row by row against `ec_data_N100000_ap1229.csv` when that optional dataset is present.

```python
import csv
from pathlib import Path


def check_csv(path="ec_data_N100000_ap1229.csv"):
    path = Path(path)
    print(f"=== Dataset sanity check: {path} ===")
    if not path.exists():
        print("Dataset not found; skipping this optional check.")
        return True

    checks = {
        "w1": 0,
        "w2": 0,
        "w3 (pipeline)": 0,
        "w3 (true w2)": 0,
    }
    mismatches = []
    n = 0

    with path.open(newline="") as f:
        for row in csv.DictReader(f):
            a2 = int(row["a_0002"])
            a3 = int(row["a_0003"])
            N = int(row["conductor"])
            w1, w2, w3 = (int(row[key]) for key in ("wa1", "wa2", "wa3"))

            predicted_w1 = compute_w1(a2)
            predicted_w2 = compute_w2(a2, a3)
            predicted_w3 = compute_w3(N, a2, predicted_w2)
            predicted_w3_from_true_w2 = compute_w3(N, a2, w2)
            results = {
                "w1": predicted_w1 == w1,
                "w2": predicted_w2 == w2,
                "w3 (pipeline)": predicted_w3 == w3,
                "w3 (true w2)": predicted_w3_from_true_w2 == w3,
            }
            for name, correct in results.items():
                checks[name] += correct
                if not correct and len(mismatches) < 10:
                    mismatches.append((row["isog_label"], name))
            n += 1

    print(f"rows: {n}")
    for name, correct in checks.items():
        print(f"{name:<18}: {correct} / {n}")

    ok = all(correct == n for correct in checks.values())
    assert ok, f"Sample dataset mismatches (isogeny class, check): {mismatches}"
    print("All formulas verified on every dataset row.")


check_csv()
```
